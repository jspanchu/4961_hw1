```python
from sklearn import datasets
import numpy as np

iris = datasets.load_iris()
iris.data
```

```python
iris.feature_names
```

```python
iris.target_names
```

# target = function of feature names

### petal length = f(sepal length,h)

```python
X = iris.data[:, 0].astype(np.float32)
y = iris.data[:, 2].astype(np.float32)
idxs = np.argsort(X)
X = X[idxs]
y = y[idxs]
X, y
```

```python
# Peek at X, y
import matplotlib.pyplot as plt
plt.plot(X, y)
plt.xlabel(iris.feature_names[0])
plt.ylabel(iris.feature_names[2])
plt.show()
```

```python
import torch
device = 'cuda' if torch.cuda.is_available() else 'cpu'

model = torch.nn.Sequential(
    torch.nn.Linear(1, 16),
    torch.nn.ReLU(),
    torch.nn.Linear(16, 32),
    torch.nn.ReLU(),
    torch.nn.Linear(32, 32),
    torch.nn.ReLU(),
    torch.nn.Linear(32, 16),
    torch.nn.ReLU(),
    torch.nn.Linear(16, 1)
)
model.to(device)
optimizer = torch.optim.Adam(params=model.parameters(), lr=0.001)
loss_fn = torch.nn.MSELoss(reduction='sum')

train_dataset = torch.utils.data.TensorDataset(
    torch.tensor(X, device=device).unsqueeze(1),
    torch.tensor(y, device=device).unsqueeze(1))
loader = torch.utils.data.DataLoader(train_dataset, batch_size=16, shuffle=True, generator=torch.Generator(device=device))

model.train()
for epoch in range(2000):
  for X_batch, y_batch in loader:
    y_pred = model(X_batch)
    loss = loss_fn(y_pred, y_batch)

    loss.backward()
    optimizer.step()
    optimizer.zero_grad()
```

```python
model.eval()

def compute_prediction(inputs):
    return model(torch.tensor(inputs).unsqueeze(1)).detach().cpu().numpy()

predictions = compute_prediction(X)
plt.plot(X, y, label='data')
plt.plot(X, predictions, color='red', label='model')
plt.xlabel(iris.feature_names[0])
plt.ylabel(iris.feature_names[2])
plt.legend()
```

```python
# Mean Squared Error (MSE)
mse_loss = np.mean((predictions - y) ** 2)

# Root Mean Squared Error (RMSE)
rmse_val = np.sqrt(mse_loss)

# Mean Absolute Error (MAE)
mae_val = np.mean(np.abs(predictions - y))

print(f"MSE: {mse_loss.item():.4f}")
print(f"RMSE: {rmse_val.item():.4f}")
print(f"MAE: {mae_val.item():.4f}")
```
